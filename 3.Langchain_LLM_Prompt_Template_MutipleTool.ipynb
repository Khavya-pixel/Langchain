```python
import os
from dotenv import load_dotenv

# Gemini
from langchain_google_genai import ChatGoogleGenerativeAI

# Modern LangChain
from langchain_core.prompts import PromptTemplate
from langchain_core.tools import Tool


# ============================================================
# 1. Load API key
# ============================================================

load_dotenv(".env")

google_api_key = os.getenv("GOOGLE_API_KEY")

print("API key loaded:", bool(google_api_key))
print("API key length:", len(google_api_key) if google_api_key else 0)


# ============================================================
# 2. Initialize Gemini
# ============================================================

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=google_api_key,
    temperature=0
)


# ============================================================
# 3. Simple QA Tool
# ============================================================

qa_prompt = PromptTemplate.from_template(
    "Answer clearly and accurately:\n\n{question}"
)

# Modern LangChain syntax
qa_chain = qa_prompt | llm


def qa_function(question):
    response = qa_chain.invoke({
        "question": question
    })

    return response.content


qa_tool = Tool(
    name="Simple QA",
    func=qa_function,
    description="Answer factual questions clearly"
)


# ============================================================
# 4. Summarizer Tool
# ============================================================

summary_prompt = PromptTemplate.from_template(
    "Summarize the following text clearly and concisely:\n\n{text}"
)

# Modern LangChain syntax
summary_chain = summary_prompt | llm


def summary_function(text):
    response = summary_chain.invoke({
        "text": text
    })

    return response.content


summary_tool = Tool(
    name="Summarizer",
    func=summary_function,
    description="Summarizes input text"
)


# ============================================================
# 5. Test QA Tool
# ============================================================

qa_query = "What is LangGraph in LangChain?"

print("\n🧠 Simple QA Tool Output:\n")

qa_result = qa_tool.run(qa_query)

print(qa_result)


# ============================================================
# 6. Test Summarizer Tool
# ============================================================

summary_text = """
LangGraph is a library built on top of LangChain that helps developers
create stateful, multi-step agents as graphs. Each node represents a step
like calling an LLM or a tool. It's ideal for advanced AI workflows.
"""

print("\n📝 Summarizer Tool Output:\n")

summary_result = summary_tool.run(summary_text)

print(summary_result)
```
